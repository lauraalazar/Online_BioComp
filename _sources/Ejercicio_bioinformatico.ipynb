## Ejercicio Bioinformático: 
#### Búsqueda y análisis de un gen de resistencia al glifosato en una especie no modelo

🔬🧪 Contexto biológico
El gen EPSPS (5-enolpiruvilshikimato-3-fosfato sintasa) es clave en la ruta del shikimato, necesaria para la biosíntesis de aminoácidos aromáticos en plantas, bacterias y hongos. El glifosato, un herbicida ampliamente usado, inhibe esta enzima.

Ciertas variantes del gen EPSPS, como las presentes en la bacteria *Agrobacterium sp.* CP4, confieren resistencia al glifosato y han sido patentadas y utilizadas en cultivos genéticamente modificados (GM).

En este ejercicio, partimos del EPSPS de una especie modelo (*Agrobacterium sp.*) y queremos identificar genes homólogos en una especie bacteriana no modelo del suelo, potencialmente con funciones similares.

### Objetivo
Investigar si una especie bacteriana del suelo (no modelo) contiene una variante del gen EPSPS que podría conferir resistencia al glifosato, y por lo tanto generar mecanismos independiente de residtencia a los del proveedor con patente, Bayer INc.

### Flujo de trabajo y herramientas
1. 🔎 Obtener la secuencia del gen EPSPS: Buscar en NCBI la secuencia del gen aroA (EPSPS) de *Agrobacterium sp.* CP4.

Descargar la secuencia nucleotídica y proteica en formato FASTA.

2. 🧬 BLAST
Usar BLASTp o tBLASTn en línea o localmente para buscar genes similares en genomas de bacterias del suelo (ej: Bacillus, Pseudomonas, Streptomyces).

Obtener las secuencias de los mejores hits.

3. 📊 Análisis de semejanza
Evaluar similitud (% identidad, cobertura, E-value) de los hits.

Elegir 3–5 secuencias candidatas para análisis posterior.

4. 🧫 Predicción de funcionalidad
Usar Pfam o HMMER para determinar si las secuencias contienen el dominio típico EPSPS (PF00275).

Verificar la conservación funcional de residuos clave.

5. 🧪 Diseño de primers
Para las secuencias candidatas, diseñar primers específicos usando herramientas como PrimerBlast. Verificar especificidad con BLAST.

6. 🌳 Análisis filogenético
Alinear las secuencias de EPSPS (referencia + candidatos + ortólogos conocidos) usando Clustal o R.

Convertir el alineamiento a phyDat (R/phangorn) y construir árboles con:

Neighbor Joining

UPGMA

ML (JTT, WAG, etc.)

Interpretar la relación evolutiva de los genes candidatos con el gen de Agrobacterium.

### Resultado esperado
1. Un árbol filogenético con los genes candidatos.

2. Reporte de similitud y conservación funcional.

3. Primers para amplificar el gen homólogo en la especie no modelo.

4. Discusión: ¿es plausible que esta bacteria tenga resistencia natural al glifosato?
