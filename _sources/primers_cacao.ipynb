### Variante de un gen

Vamos a estudiar una secuencias de genes codificantes para transportadores NRAMP5 de una variedad colombiana (CCN51) con el genoma de referencia disponible en NCBI (NCBI RefSeq assembly: GCF_000208745.1), para identificar similitudes o diferencias, localizar los genes en los cromosomas, y diseñar primers para PCR convencional. 


1. Vamos a descargar la secuencia https://raw.githubusercontent.com/Pipe-95/TallerBiologiaComputacional/refs/heads/main/NRAMP5.fa

2. Ahora descargamos los cromosomas dada la lista de ID de los cromosomas

```python
import Bio
from Bio import Entrez, SeqIO

Entrez.email = "farangoh@eafit.edu.co" #modifique el email con el suyo

# Lista de IDs de los cromosomas a descargar
cromosomas_ids = ["NC_030850.1", "NC_030851.1", "NC_030852.1", "NC_030853.1", "NC_030854.1", "NC_030855.1", "NC_030856.1", "NC_030857.1", "NC_030858.1", "NC_030859.1"]

# Ruta del archivo final fusionado
ruta_salida = "genoma_cacao.txt"

# Lista para guardar las secuencias
secuencias = []

# Descargar y almacenar cada secuencia
for idx, crom_id in enumerate(cromosomas_ids, start=1): 
    try:
        with Entrez.efetch(db="nucleotide", rettype="fasta", retmode="text", id=crom_id) as handle:
            seq_record = SeqIO.read(handle, "fasta")
            secuencias.append(seq_record) # append: añade un item a la lista en vez de sobreescribir
            print(f"Cromosoma {idx} guardado exitosamente: ID {seq_record.id}, longitud {len(seq_record.seq)} pb")
    except Exception as e:
        print(f"Error al descargar cromosoma {idx} (ID {crom_id}): {e}")


# Escribir todas las secuencias en un solo archivo FASTA
try:
    SeqIO.write(secuencias, ruta_salida, "fasta")
    print(f"\nTodas las secuencias fueron fusionadas en: {ruta_salida}")
except Exception as e:
    print(f"Error al escribir las secuencias")
```

3. Generar la base de datos para BLAST utilizand mkblastdb (dentro de un script slurm)

```python
##### ENVIRONMENT CREATION #####
module load blast/2.13.0_gcc-11.2.0

##### JOB COMMANDS ####
makeblastdb -in genoma_cacao.txt -title Base_datos_T.cacao -dbtype nucl -out cacao_database -parse_seqids
```

4. Ahora realizamos Blast de la secuencia query (NRAMP5) contra la base de datos cacao_database

```python
##### ENVIRONMENT CREATION #####
module load blast/2.13.0_gcc-11.2.0

blastn -query NRAMP5.fa -db cacao_database -out NRAMP5_blastn_table.txt -evalue 1e-30 -outfmt 6
```
