## Cómo correr programas en Apolo

En Apolo se deben correr programas por medio de scripts, ya que no hay una interfase gráfica. Algunos programas que utilizamos están instalados, sin embargo a veces necesitamos algunas versiones específicas u otras librerías y para esto debemos cargar el módulo correspondientes. Para ver la disponibilidad de módulos utilizamos:

```python
module avail
```

Para obtener los programas necesarios para correr los scripts, se necesita instalarlos por medios de ambientes conda. Este es un gestor de programas que permite que haya múltiples versiones del programa y que se instale con todo lo necesario

```python

module load python/3.6.0_miniconda-4.3.11_gcc-11.2.0
conda create -n biopython
conda activate biopython
conda install -c conda-forge biopython
conda install -c conda-forge pandas
```

Debemos salir y volver a entrar a apolo para chequear la instalación. Una vez realizamos el login, activamos el ambiente de bioppython y comenzamos python

```python
source activate biopython
python
```

```python
import Bio
print(Bio.__version__)
```

Si sale un error después de llamar la librería es porque hay un error de instalación y hay que repetirla.

## Correr programas por medio de un script SLURM

Un script slurm es un script ejecutable en bash (``.sh'') tiene la siguiente estructura:
1. Un encabezado con el usuario 
2. Parámetros de SBATCH: aquí se especifica: la partición (learning en neustro caso), los nodos a usar, el tiempo, el email del usuario
3. El cuerpo del script: aquí se deben cargar los módulos, activar los ambientes e insertar el código a ejecutar


#!/bin/bash
#==============================================================================
# ==============================================================================
#==============================================================================
#   Slurm Job Submission Script
#
#   Generated by:   Interactive Slurm Script Generator v1.0.1
#   Author:         lsalazarj
#   Generated On:   Tue Jul 29 15:44:17 -05 2025
#
# ==============================================================================
#

#==============================================================================
# SBATCH DIRECTIVES (generated on Tue Jul 29 15:44:59 -05 2025)
#==============================================================================
#SBATCH --job-name="ecoli_mutantes"
#SBATCH --partition=learning
#SBATCH --nodes=1
#SBATCH --ntasks=1
#SBATCH --mem=4G
#
#SBATCH --time=00:10:00
#SBATCH --output=logs/%x_%j.out      # Standard output log (%x = job name, %j = job ID)
#SBATCH --error=logs/%x_%j.err       # Standard error log
#SBATCH --mail-type=ALL              # Mail events (NONE, BEGIN, END, FAIL, ALL)
#SBATCH --mail-user=lsalazarj@eafit.edu.co         # Your email address from user input

#==============================================================================
# SCRIPT BODY
#==============================================================================

echo "================================================="
echo "Starting job $SLURM_JOB_ID on host $(hostname)"
echo "Job name: $SLURM_JOB_NAME"
echo "Partition: $SLURM_JOB_PARTITION"
echo "Number of nodes: $SLURM_NNODES"
echo "Total number of tasks: $SLURM_NTASKS"
echo "================================================="
echo

# --- Environment Setup ---
# Load any necessary modules
module load python/3.12_miniconda-24.7.1

# Activate a virtual environment
source activate biopython


# --- Application Execution ---
echo "Starting application..."
echo

# Insert code here

python ExamenParcial.py

echo
echo "================================================="
echo "Job finished with exit code $?"
echo "================================================="
