# Ejemplo de script ejecutable en bash 

## Automatizar y documentar un script en bash
1. Vamos a escribir un script *ejecutable* en bash (por lo tanto su extensión será ``.sh``) que busca contar las secuencias presentes en el archivo descargado "Sars_cov.prot.fa"

```python
# primero accedemos al directorio donde está el archivo que queremos usar de input
cd sars_covid19
```

2. Utilizamos un editor de texto para escribir el script

```python
nano contador_secuencias.sh
```

```python
#!/bin/bash

# Definir la variable con el nombre del archivo
fasta_file="Sars_cov.prot.fa"

# Verificar si el archivo existe
if [ ! -f "$fasta_file" ]; then
    echo "El archivo $fasta_file no existe."
    exit 1
fi

# Contar líneas que empiezan con '>'
sequence_count=$(grep -c "^>" "$fasta_file")

# Mostrar resultados
echo "Archivo analizado: $fasta_file"
echo "Número de secuencias encontradas: $sequence_count"
```

3. Ejecutar el archivo: esto se puede hacer de dos formas
    - Utilizando el comando del programa

```python

bash contador_secuencias.sh
```

- Haciéndolo ejecutable

```python
chmod +x contador_secuencias.sh
./contador_secuencias.sh
```
